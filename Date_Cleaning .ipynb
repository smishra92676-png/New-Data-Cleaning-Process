# Handle Missing Value 

```python
df.isnull().sum()
```

```python
df.dropna(subset=['Customer_ID'],inplace=True)
```

```python
df['Age'].unique()
```

```python

df['Age'] = df['Age'].replace('nan', pd.NA)

df['Age'] = pd.to_numeric(
    df['Age'].astype(str).str.extract(r'(\d+)')[0],
    errors='coerce'
)

age_median = int(df['Age'].median())

df['Age'] = df['Age'].fillna(age_median)

df['Age'] = df['Age'].astype(int)
```

```python
df['Age'].unique()
```

```python
df['Purchase_Amount']=df['Purchase_Amount'].fillna(df['Purchase_Amount'].mean())
```

```python
df['Feedback_Score'].fillna(df['Feedback_Score'].mode()[0],inplace=True)
```

```python
df.head(2)
```

```python
df.isnull().sum()
```

```python
df['Feedback_Score'].unique()
```

```python
for col in ['Gender','City','Country']:
    df[col].fillna(df[col].mode()[0],inplace=True)
```

```python
df['Last_Purchase_Date'] = df['Last_Purchase_Date'].ffill()
```
